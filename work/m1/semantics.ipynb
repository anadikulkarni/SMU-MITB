Double equals == cannot be used for floats as a lot of floats are actually approximations of the real number. Use the decimal library: decimal.Decimal(0.1) + decimal.Decimal(0.2) == decimal.Decimal(0.3) or math.isclose() to compare 2 float 
values properly.

```python
import decimal
decimal.Decimal(0.1) + decimal.Decimal(0.2) == decimal.Decimal(0.3)
```

```python
decimal.Decimal("0.1") + decimal.Decimal("0.2") == decimal.Decimal("0.3")
```
